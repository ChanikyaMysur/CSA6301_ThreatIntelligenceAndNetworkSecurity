# Experiment 1: Security Control Baseline Checker
**Topic:** Network Security Fundamentals
**Aim:** Build a tool that checks an organization's current security control configuration against a baseline.


```python
def evaluate_security_baseline(config, baseline):
    """
    config: {control_name: bool} current state.
    baseline: {control_name: bool} required minimum (True = mandatory).
    Returns the list of controls that fail to meet the baseline.
    """
    failures = []
    for control, required in baseline.items():
        current = config.get(control, False)
        if required and not current:
            failures.append(control)
    return failures
```

```python
def test_experiment1():
    baseline = {
        "firewall_enabled": True,
        "password_policy_enforced": True,
        "patches_up_to_date": True,
        "mfa_enabled": True,
    }
    config = {
        "firewall_enabled": True,
        "password_policy_enforced": True,
        "patches_up_to_date": False,
    }
    failures = evaluate_security_baseline(config, baseline)
    assert "patches_up_to_date" in failures
    assert "mfa_enabled" in failures
    assert "firewall_enabled" not in failures
    assert "password_policy_enforced" not in failures
    assert len(failures) == 2
    print("All test cases passed.")

test_experiment1()
```
