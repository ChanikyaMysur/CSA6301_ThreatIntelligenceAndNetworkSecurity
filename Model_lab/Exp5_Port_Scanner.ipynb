# Experiment 5: Simple Vulnerability Assessment (Port Scanner)
**Aim:** Perform vulnerability assessment by scanning open ports on a target host.


```python
import socket

def scan_ports(target_host, ports):
    open_ports = []
    for port in ports:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(0.5)
        result = sock.connect_ex((target_host, port))
        if result == 0:
            open_ports.append(port)
        sock.close()
    return open_ports
```

```python
target = "127.0.0.1"
ports_to_check = [21, 22, 80, 443, 8080]
print("Open ports:", scan_ports(target, ports_to_check))
```
