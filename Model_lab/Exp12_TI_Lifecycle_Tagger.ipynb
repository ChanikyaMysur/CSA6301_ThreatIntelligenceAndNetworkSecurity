# Experiment 12: Threat Intelligence Lifecycle Tagger
**Aim:** Map security operational activities to stages of the Threat Intelligence Lifecycle.


```python
def tag_lifecycle_stage(activity_description):
    d = activity_description.lower()
    if any(k in d for k in ["requirement", "plan", "scope", "goal"]):
        return "Direction"
    elif any(k in d for k in ["scraping", "gather", "log capture", "pull"]):
        return "Collection"
    elif any(k in d for k in ["parsing", "normalize", "decrypt", "clean"]):
        return "Processing"
    elif any(k in d for k in ["correlate", "find pattern", "attribution", "examine"]):
        return "Analysis"
    elif any(k in d for k in ["publish", "briefing", "report", "alert"]):
        return "Dissemination"
    elif any(k in d for k in ["review", "feedback", "improvement", "survey"]):
        return "Feedback"
    return "Unknown"
```

```python
desc = "Normalizing raw XML threat feeds into JSON format."
print("Lifecycle Stage:", tag_lifecycle_stage(desc))
```
