# Experiment 10: Mini ELK Stack Simulation
**Aim:** Model Logstash parsing, Elasticsearch indexing, and Kibana aggregations.


```python
def run_elk_pipeline():
    # 1. Logstash: Parse raw log line into structured JSON
    raw_log = "2026-08-25 10:15:30 status=404 user=admin"
    parts = raw_log.split()
    parsed_event = {"timestamp": f"{parts[0]} {parts[1]}", "status": parts[2].split("=")[1], "user": parts[3].split("=")[1]}

    # 2. Elasticsearch: Index parsed document into list store
    index = []
    index.append(parsed_event)

    # 3. Kibana: Aggregation query
    aggregation = {}
    for doc in index:
        status = doc["status"]
        aggregation[status] = aggregation.get(status, 0) + 1

    print("Parsed & Indexed Event:", index[0])
    print("Kibana Aggregation:", aggregation)
```

```python
run_elk_pipeline()
```
