# Experiment 4: Threat Intelligence Application Area Classifier
**Aim:** Identify relevant threat intelligence application areas for business scenarios.


```python
def classify_ti_application(scenario):
    s = scenario.lower()
    if any(k in s for k in ["patient", "hospital", "hipaa", "medical"]):
        return "Healthcare Data Protection"
    elif any(k in s for k in ["card", "bank", "atm", "swift", "transaction"]):
        return "Banking Fraud Prevention"
    elif any(k in s for k in ["cart", "checkout", "store", "order"]):
        return "E-Commerce Fraud Prevention"
    return "General Threat Intelligence"
```

```python
scenario = "Detecting credit card cloning at local ATM terminals."
result = classify_ti_application(scenario)
print(f"Result: {result}")
```
