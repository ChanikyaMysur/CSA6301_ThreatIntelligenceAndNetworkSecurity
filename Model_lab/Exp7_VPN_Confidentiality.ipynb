# Experiment 7: VPN Confidentiality Simulation
**Aim:** Model confidentiality in transit using a symmetric cipher simulation.


```python
def xor_encrypt_decrypt(data_bytes, key_byte):
    return bytes([b ^ key_byte for b in data_bytes])
```

```python
plaintext = b"CONFIDENTIAL_PAYROLL_DATA"
key = 0xAF

ciphertext = xor_encrypt_decrypt(plaintext, key)
print("Captured in transit (Unreadable):", ciphertext)

decrypted = xor_encrypt_decrypt(ciphertext, key)
print("Decrypted with shared key:", decrypted.decode())
```
