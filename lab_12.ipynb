```python
from datetime import datetime

events = [
    {"time": "09:10", "type": "Failed Login", "severity": "Medium"},
    {"time": "09:12", "type": "Failed Login", "severity": "Medium"},
    {"time": "09:15", "type": "Port Scan Detected", "severity": "High"},
    {"time": "09:30", "type": "Malware Detected", "severity": "Critical"}
]

report_name = "incident_report.txt"

with open(report_name, "w") as report:

    report.write("========== INCIDENT REPORT ==========\n")
    report.write(f"Generated: {datetime.now()}\n\n")

    report.write("Security Events:\n")
    report.write("----------------------------\n")

    for i, event in enumerate(events, 1):
        report.write(f"{i}. Time: {event['time']}\n")
        report.write(f"   Event: {event['type']}\n")
        report.write(f"   Severity: {event['severity']}\n\n")

    report.write("----------------------------\n")
    report.write(f"Total Events: {len(events)}\n")

print("Incident report generated successfully!")
print("Saved as:", report_name)
```
