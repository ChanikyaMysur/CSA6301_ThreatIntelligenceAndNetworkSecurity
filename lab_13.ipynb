```python
import hashlib

def calculate_hash(filename):
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)

            if not data:
                break

            sha256.update(data)

    return sha256.hexdigest()


filename = input("Enter file name: ")

# Create baseline hash
baseline_hash = calculate_hash(filename)

print("\nBaseline SHA-256 Hash:")
print(baseline_hash)

input("\nModify the file if desired, then press Enter to continue...")

current_hash = calculate_hash(filename)

print("\nCurrent SHA-256 Hash:")
print(current_hash)

if baseline_hash == current_hash:
    print("\nFile Integrity Verified.")
    print("No modifications detected.")
else:
    print("\nWARNING!")
    print("File has been modified.")
```
