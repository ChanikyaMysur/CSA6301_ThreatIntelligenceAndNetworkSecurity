# Experiment 2: CIA Triad Violation Detector
**Topic:** Security Goals (CIA Triad)
**Aim:** Detect which CIA principles are violated based on system security properties.


```python
def classify_cia_violations(properties):
    """
    properties: list of (description, category, status). Returns a dict
    {category: [violated_descriptions]} built from every property whose status is False.
    """
    violations = {}
    for desc, category, status in properties:
        if not status:
            violations.setdefault(category, []).append(desc)
    return violations
```

```python
def test_experiment2():
    properties = [
        ("Data encrypted at rest", "Confidentiality", True),
        ("Role-based access control enforced", "Confidentiality", False),
        ("File checksums verified after transfer", "Integrity", True),
        ("Audit logs digitally signed", "Integrity", False),
        ("Redundant server cluster with failover", "Availability", True),
    ]
    violations = classify_cia_violations(properties)
    assert "Confidentiality" in violations
    assert "Role-based access control enforced" in violations["Confidentiality"]
    assert "Integrity" in violations
    assert "Audit logs digitally signed" in violations["Integrity"]
    assert "Availability" not in violations
    print("All test cases passed.")

test_experiment2()
```
