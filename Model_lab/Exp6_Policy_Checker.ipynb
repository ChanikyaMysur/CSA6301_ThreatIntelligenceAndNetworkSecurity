# Experiment 6: Security Governance Policy Checker
**Aim:** Audit security policy drafts for missing required governance elements.


```python
def check_policy(document_text):
    text = document_text.lower()
    required = ["scope", "roles and responsibilities", "enforcement", "review cycle", "approval authority"]
    missing = [elem for elem in required if elem not in text]
    return missing
```

```python
policy = "This policy defines the Scope and Enforcement rules, with annual Review Cycle."
print("Missing Elements:", check_policy(policy))
```
