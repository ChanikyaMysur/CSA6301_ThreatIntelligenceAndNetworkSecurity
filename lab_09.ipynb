```python
# ---------------------------------------------
# Firewall Simulation in Python
# ---------------------------------------------

# Function to check whether a packet is allowed or blocked
def firewall(packet):

    # Rule 1: Block packets from blacklisted IP
    if packet["source_ip"] == "192.168.1.100":
        return "BLOCKED - Blacklisted Source IP"

    # Rule 2: Block packets going to port 23 (Telnet)
    elif packet["destination_port"] == 23:
        return "BLOCKED - Telnet Port Not Allowed"

    # Rule 3: Allow HTTP traffic (Port 80)
    elif packet["destination_port"] == 80:
        return "ALLOWED - HTTP Traffic"

    # Rule 4: Allow HTTPS traffic (Port 443)
    elif packet["destination_port"] == 443:
        return "ALLOWED - HTTPS Traffic"

    # Rule 5: Block all other traffic
    else:
        return "BLOCKED - Unknown or Restricted Port"


# Sample network packet
packet = {
    "source_ip": "192.168.1.10",
    "destination_ip": "10.0.0.5",
    "destination_port": 80,
    "protocol": "TCP"
}

# Display packet information
print("Packet Details")
print("---------------------------")
print("Source IP      :", packet["source_ip"])
print("Destination IP :", packet["destination_ip"])
print("Port           :", packet["destination_port"])
print("Protocol       :", packet["protocol"])

# Firewall decision
result = firewall(packet)

print("\nFirewall Decision:")
print(result)
```
