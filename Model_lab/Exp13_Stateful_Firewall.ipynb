# Experiment 13: Stateful Inspection Firewall Simulator
**Aim:** Simulate a stateful firewall that tracks connection states.


```python
class StatefulFirewall:
    def __init__(self):
        self.state_table = set()

    def send_outbound(self, src_ip, src_port, dst_ip, dst_port):
        connection = (src_ip, src_port, dst_ip, dst_port)
        self.state_table.add(connection)
        return "Outbound Traffic Permitted"

    def receive_inbound(self, src_ip, src_port, dst_ip, dst_port):
        expected_outbound = (dst_ip, dst_port, src_ip, src_port)
        if expected_outbound in self.state_table:
            return "Inbound Traffic Permitted (Established State)"
        return "Inbound Traffic BLOCKED (Unsolicited Packet)"
```

```python
fw = StatefulFirewall()
print(fw.send_outbound("192.168.1.5", 49152, "8.8.8.8", 80))
print(fw.receive_inbound("8.8.8.8", 80, "192.168.1.5", 49152))
print(fw.receive_inbound("45.33.32.156", 443, "192.168.1.5", 49152))
```
