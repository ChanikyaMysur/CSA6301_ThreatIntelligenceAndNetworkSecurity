```python
# Basic Signature-Based Intrusion Detection System (IDS)
# Detects repeated failed login attempts from the same IP

THRESHOLD = 3

failed_attempts = {}

print("=== Basic Signature-Based IDS ===")
print("Enter login attempts.")
print("Type 'exit' as IP to stop.\n")

while True:
    ip = input("Enter IP Address: ")

    if ip.lower() == "exit":
        break

    status = input("Login Status (success/failure): ").lower()

    if status == "failure":
        failed_attempts[ip] = failed_attempts.get(ip, 0) + 1

        print(f"Failed attempts from {ip}: {failed_attempts[ip]}")

        if failed_attempts[ip] >= THRESHOLD:
            print("*************************************")
            print("ALERT! Possible Intrusion Detected!")
            print(f"IP Address: {ip}")
            print(f"Failed Attempts: {failed_attempts[ip]}")
            print("*************************************")

    elif status == "success":
        # Reset counter after successful login
        failed_attempts[ip] = 0
        print("Successful login. Failure count reset.")

    else:
        print("Invalid login status. Use 'success' or 'failure'.")

print("\nMonitoring Stopped.")
```
