# Experiment 3: Network Threat Type Classifier
**Topic:** Types of Network Threats
**Aim:** Build a keyword-based classifier that identifies threat categories from incident descriptions.


```python
THREAT_RULES = [
    ("Phishing", ["click the link", "verify your account", "password reset"]),
    ("DoS/DDoS", ["unreachable", "overwhelm", "syn packets"]),
    ("Malware", ["encrypted", "ransom", "malicious file"]),
    ("Man-in-the-Middle", ["arp", "intercept"]),
]

def classify_threat(description, rules=THREAT_RULES):
    """Return the first matching threat category, or 'Unknown' if no keyword matches."""
    d = description.lower()
    for category, keywords in rules:
        if any(k in d for k in keywords):
            return category
    return "Unknown"
```

```python
def test_experiment3():
    cases = [
        ("An attacker sent an email asking the victim to click the link to verify your account details.", "Phishing"),
        ("The server became unreachable after being hit by an overwhelming volume of SYN packets.", "DoS/DDoS"),
        ("Files on the workstation were encrypted and a ransom note appeared.", "Malware"),
        ("The attacker used ARP poisoning to intercept traffic between two hosts.", "Man-in-the-Middle"),
        ("The office coffee machine needs descaling.", "Unknown"),
    ]

    for description, expected in cases:
        result = classify_threat(description)
        assert result == expected, f"Expected {expected}, got {result} for: {description}"
    print("All test cases passed.")

test_experiment3()
```
