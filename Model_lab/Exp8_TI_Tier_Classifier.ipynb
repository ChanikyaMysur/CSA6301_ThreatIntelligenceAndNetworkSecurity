# Experiment 8: Threat Intelligence Tier Classifier
**Aim:** Classify intelligence reports as Strategic, Tactical, or Operational.


```python
def classify_ti_tier(report_text, audience):
    combined = (report_text + " " + audience).lower()
    if any(k in combined for k in ["ceo", "ciso", "board", "executive", "budget"]):
        return "Strategic"
    elif any(k in combined for k in ["ip", "hash", "ioc", "soc", "analyst"]):
        return "Operational"
    elif any(k in combined for k in ["ttp", "malware", "vector", "technique"]):
        return "Tactical"
    return "Operational"
```

```python
report = "List of malicious IP indicators"
audience = "SOC Analysts"
print("Tier:", classify_ti_tier(report, audience))
```
