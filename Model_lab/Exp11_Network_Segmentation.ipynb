# Experiment 11: Network Segmentation Lateral Movement Model
**Aim:** Model network segment ACL policies to block lateral movement.


```python
SEGMENT_RULES = {
    "DMZ": ["Internal_App"],
    "Internal_App": ["Database"],
    "Guest_WiFi": []
}

def can_move_laterally(source_segment, target_segment):
    allowed = SEGMENT_RULES.get(source_segment, [])
    return target_segment in allowed
```

```python
print("Guest_WiFi -> Database Access:", can_move_laterally("Guest_WiFi", "Database"))
print("Internal_App -> Database Access:", can_move_laterally("Internal_App", "Database"))
```
